### File Handling:
                    The file is used to store data in electronics formate in system

            File operations:
                            1.Opening a file---Open()
                            2.Reading a file---Read()
                            3.Writing into a file--Write()
                            4.Closing a file----close()
                            
            1.Opening a file:
                            To open a file by using open() in that ,we need to pass file name and file opening mode

                        Syntax:
                            file_obj=open(file_name,mode)
                            # read/write operations
                            # close file

           1.File opening modes:
                                1. r-->(read)--->Opening file in read mode.(if file not exit it raise "FileNotFoundError"
                                2. w-->(write)--->Opening a file in write mode.(if file is not exit it create new file otherwise it erase the previous contents)
                                3. a-->(append)--->Opening a file in append mode.(if file not exits it create new file otherwise it add content to the previous contents)
                                4. x-->(create)--->Create new file
                                5. b-->(binary)--->rb(read binary),wb(write binary)-->This mode is used to performed operation on binary files for writting "wb" mode for read "rb" mode.
                                6. r+-->read+write
                                7. w+-->write+read
                                8. a+-->append+read

            Note:

                    Syntax:
                            with open(file_name,mode) as file_obj:
                                    #file operations
                            

            2.Reading content from a file:
                                            To perform the read operations on file for that we need open in "r" mode.and then we following operations.


                                1.read():It return entire file contents as single string.
                                2.readline():This methods return first line.
                                3.readlines():It return list of lines.

            3.Writing into a file:If you open a file in write mode we can perform following operations.

                                1.Write(String):It add string into a file.
                                2.Writelines(List_of_lines):It adds all list lines to a file.


            4.Closing a file:After perform read/write operation an a file,we need to close the file by using close function otherwise data loss will happend.
