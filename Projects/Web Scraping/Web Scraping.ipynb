```python

search_ele =input("Enter product name:")
url = "https://www.flipkart.com/search?q=" + search_ele.replace(" ","")
try:
    headers = {
    "User-Agent": "Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/138.0.0.0 Safari/537.36",
    "Accept-Language": "en-US,en;q=0.9",
    "Accept": "text/html,application/xhtml+xml"
    }
    web_response = requests.get(url,headers=headers)
    print(web_response)

    # converting text into html parser
    web_content =bs(web_response.text,'html.parser')
    products = web_content.find_all('div',{'class':"lvJbLV col-12-12"})

    # iterate each product 

    for product in products:
        try:
            product_link ="https://www.flipkart.com" + product.div.div.div.a['href']
        except:
            product_link ="No product Link"
    
            # product name
        try:
            product_name = product.div.div.div.a.find('div',{"class":"HZ0E6r Rm9_cy"}).text
        except:
            product_name = "No product Name"
    
            #  Actual price
        try:
            Actual_price = product.div.div.div.a.find('div',{"class":"hZ3P6w DeU9vF"}).text
        except:
            Actual_price = "No product price"
    
            # Rating
        try:
            product_rating = product.div.div.div.a.find('div',{'class':"MKiFS6"}).text
        except:
            product_rating = "No product rating"
    
    
            # Description
        try:
            product_des_tag=product.div.div.div.a.find_all('div', {'class':'CMXw7N'})
        except:
            product_des_tag="no product_des_tag"
            
    
            # product info
    
        product_info={
            'product Name':product_name,
            'Actual price':Actual_price,
            'Description':product_des_tag,
            'Rating':product_rating,
            'Product Link':product_link
        }
        print(product_info)
        print("*" * 100)
except Exception as e:
    print("Side don't allow scraping:", e)
```
